```python
# create model class -> binary classifier with 5 inputs
import torch
import torch.nn as nn

class Model(nn.Module):

    def __init__(self, num_features):

        super().__init__()
        # add all layers of our network in a Sequential container
        self.network = nn.Sequential(
            nn.Linear(num_features, 3),
            nn.ReLU(),
            nn.Linear(3, 1),
            nn.Sigmoid()
        )

    # this function will be called whenever an instance of a class is called by providing the parameters
    # this method is called inside __call__() magic function
    def forward(self, features): 

        out = self.network(features)
        
        return out
```

```python
# create dataset
features = torch.rand(10, 5)

# create model
model = Model(features.shape[1])

# call model for forward pass
model(features) # same as model.forward(features)
```
